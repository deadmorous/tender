# A guided derivation

Choosing the next step in a derivation is the hard part: `tender` has forty of
them, and the one you want depends on the expression in front of you.
`td.explore` turns that recall exercise into a click.

**This notebook needs a live kernel** — unlike the other examples it produces no
LaTeX file, because its output *is* the interaction.  It also needs
`ipywidgets`, which is optional for the rest of the library:

```bash
pip install ipywidgets
```


````python
import tender
import tender.basis as tb
import tender.derivation as td
import tender.steps as ts
````

## The preamble is the cell you already wrote

The session takes the objects the steps need from *this* namespace, by kind.
So the frame below is named once, here, and never again — not in a call, not in
a text field.

````python
ctx = tender.Context()
frame = tb.wcs(ctx)

a = tender.tensor("a", rank=1, ctx=ctx)
b = tender.tensor("b", rank=1, ctx=ctx)

start = a % (a % b)          # a × (a × b)
start.latex()
````

## Open the surface

Under each expression is a chooser holding the steps that *actually do
something to it* — with what each would change, so the list reads as options
rather than as a menu.  Choose one and the derivation extends; choose at an
earlier point and the tail is dropped, though the abandoned branch is kept and
a step tried before is marked `· tried`.

Steps that were **not tried** because an argument was missing are listed
separately: they are a third category, and supplying what they lack is often
the next move.

Try, for instance: `expand_in_basis` → `reduce_frame` → `contract_eps_pair` →
`reduce_frame` → `reassemble`.

````python
s = td.explore(start)
````

## The session outlives the widget

Nothing is saved and nothing is lost: `s` is an ordinary Python object, so
closing the output above decides nothing.

````python
print(s.script())          # the path so far, as code to paste
````

````python
print(s.attempts())        # the whole tree, abandoned branches included
````

## The same answers, without the widget

Everything the surface shows is a library call, so a terminal loses the clicks
and nothing else.

````python
print(s.applicable())
````

````python
print(s.why_not("contract_delta"))
````
